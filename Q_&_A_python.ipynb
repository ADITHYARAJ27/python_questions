```python
"""Temperature Conversion:
Ask the user to input a temperature in Fahrenheit and convert it to Celsius. If the Celsius temperature is below 0, print "Freezing cold"; between 0 and 20, print "Cold";
 between 20 and 30, print "Warm"; and above 30, print "Hot"."""

fahrenheit=float(input("enter the temperature"))
celsius = (fahrenheit - 32) * 5 / 9
print(celsius)
if celsius<=0:
  print("freezing cold")
elif celsius>0 and celsius<=20:
  print("cold")
elif celsius>20 and celsius<=30:
  print("warm")
elif celsius>30:
  print("hot")
```
