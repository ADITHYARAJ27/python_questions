5. Student Score Analysis

Question: Write a program to take the names and marks of students as input. Store them in a dictionary and perform the following tasks:

1. Print the name of the student with the highest marks.


2. Print the average marks of the class.



Input:

Number of students: 3
Name: Alice, Marks: 85
Name: Bob, Marks: 90
Name: Charlie, Marks: 78

Output:

Topper: Bob
Average Marks: 84.33


Hint: Use a dictionary to store names and marks. Use loops to calculate the maximum marks and the average.

```python
students = {}

n = int(input("Number of students: "))

for i in range(n):
    name = input("Name: ")
    marks = int(input("Marks: "))
    students[name] = marks

# Find topper
topper = max(students, key=students.get)

# Calculate average
total = sum(students.values())
average = total / n

print("Topper:", topper)
print("Average Marks:", round(average, 2))
```
